```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter

# ---------------------------
# 1. Import and Visualize the Graph
# ---------------------------
# NetworkX includes the Les Miserables graph as a built-in function.
G = nx.les_miserables_graph()
```

```python
# Visualize the graph using a spring layout.
plt.figure(figsize=(20, 16))
pos = nx.spring_layout(G, seed=42)  # seed for reproducibility
nx.draw_networkx(G, pos, with_labels=True, node_size=300, font_size=8, font_color="blue", node_color="red", edge_color="gray")
plt.title("Les Miserables Character Coappearance Graph")
plt.show()
```

```python
# ---------------------------
# 2. Node Centrality Measures and Top 5 Nodes
# ---------------------------
def print_top_5(centrality_dict, measure_name):
    # sort items by centrality score in descending order
    sorted_items = sorted(centrality_dict.items(), key=lambda x: x[1], reverse=True)
    print(f"Top 5 nodes for {measure_name}:")
    for node, score in sorted_items[:5]:
        print(f"  {node}: {score:.4f}")
    print("\n")
```

```python
# 2.1 Degree Centrality
deg_cent = nx.degree_centrality(G)
print_top_5(deg_cent, "Degree Centrality")
```

```python
# 2.2 Eigenvector Centrality
# using the power method built in
eigen_cent = nx.eigenvector_centrality(G, max_iter=1000)
print_top_5(eigen_cent, "Eigenvector Centrality")
```

```python
import numpy as np

# Convert the graph to a numpy array (adjacency matrix)
A = nx.to_numpy_array(G)
# Compute the eigenvalues (take absolute values in case some are complex)
eigenvalues = np.abs(np.linalg.eigvals(A))
largest_eig = np.max(eigenvalues)
print(f"Largest eigenvalue: {largest_eig:.4f}")
```

```python
# 2.3 Katz Centrality
# A typical choice for alpha is below the reciprocal of the largest eigenvalue.
# NetworkX will choose a default if not provided but here we set one explicitly.

alpha_val = 0.9 / largest_eig

katz_cent = nx.katz_centrality(G, alpha=alpha_val, beta=1.0, max_iter=1000)
print_top_5(katz_cent, "Katz Centrality")
```

```python
# 2.4 PageRank Centrality
pagerank_cent = nx.pagerank(G, tol=1e-6,alpha=0.85)
print_top_5(pagerank_cent, "PageRank Centrality")
```

```python
# 2.5 Betweenness Centrality
betweenness_cent = nx.betweenness_centrality(G)
print_top_5(betweenness_cent, "Betweenness Centrality")
```

```python
# 2.6 Closeness Centrality
closeness_cent = nx.closeness_centrality(G)
print_top_5(closeness_cent, "Closeness Centrality")
```

```python
# ---------------------------
# 3. Graph Measures
# ---------------------------
# Degree Distribution
degree_sequence = [d for n, d in G.degree()]
degree_count = Counter(degree_sequence)
# Sort the distribution by degree value
sorted_degree = sorted(degree_count.items())
print("Degree Distribution (degree: count):")
for degree, count in sorted_degree:
    print(f"  {degree}: {count}")
print("\n")
```

```python
# Global Clustering Coefficient (Transitivity)
global_clustering = nx.transitivity(G)
print(f"Global Clustering Coefficient (Transitivity): {global_clustering:.4f}")
```

```python
# Diameter: we compute diameter on the largest connected component (this will work if the graph might be disconnected)
if nx.is_connected(G):
    print("Graph is connected.")
    diameter = nx.diameter(G)
else:
    print("Graph is not connected. Computing diameter on the largest connected component.")
    largest_cc = max(nx.connected_components(G), key=len)
    subgraph = G.subgraph(largest_cc)
    diameter = nx.diameter(subgraph)
print(f"Diameter of the graph (or largest component): {diameter}\n")
```

```python
# ---------------------------
# 4. Pagerank Centrality with NetworKit
# ---------------------------

import networkx as nx
import networkit as nk

# 1. Load the Les Miserables graph using NetworkX.
G_nx = nx.les_miserables_graph()

# Create a list of original node labels (to map back later).
nx_nodes = list(G_nx.nodes())

# 2. Convert the NetworkX graph to a NetworKit graph.
G_nk = nk.nxadapter.nx2nk(G_nx)

# 3. Compute the PageRank scores using NetworKit.
pr = nk.centrality.PageRank(
    G_nk,
    tol = 0.01,
    damp = 0.85,
)

pr.run()

# 4. Retrieve the number of iterations until convergence.
iterations = pr.numberOfIterations()

# 5. Retrieve the PageRank scores.
scores = pr.scores()

# Build a mapping from NetworKit node IDs (integers) back to the original NetworkX node labels.
mapping = {i: nx_nodes[i] for i in range(len(nx_nodes))}

# Create a dictionary mapping each original node label to its PageRank score.
pr_dict = {mapping[i]: score for i, score in enumerate(scores)}

# 6. Sort the nodes by PageRank score in descending order and print the top 5.
sorted_pr = sorted(pr_dict.items(), key=lambda x: x[1], reverse=True)

print("Top 5 nodes by PageRank:")
for node, score in sorted_pr[:5]:
    print(f"  {node}: {score:.4f}")
print(f"Converged in {iterations} iterations, with tolerance 0.01.")
```
