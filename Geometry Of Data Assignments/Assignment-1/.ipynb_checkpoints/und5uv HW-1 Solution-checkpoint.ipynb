### Question 1
Consider the set of all **closed** intervals in the real line with
  non-zero length:
  $$\mathcal{B} = \{[a, b] : a < b\}.$$
  Is this a valid basis for a topology on $\mathbb{R}$? Why or why not?



### Solution
The given set $\mathcal{B}$ is not a valid basis for a topology on $\mathbb{R}$.

Proof : let's consider the following two intervals $[x,y]$ and $[y,z]$ where $x < y < z$ from the given Basis $\mathcal{B}$. in this scenario, the intersection of these two intervals has only one element $y$
  
From the definition of a **basis** for a topology, If $B_1, B_2 \in \mathcal{B}$ and $x \in (B_1 \cap B_2)$, then there exists a $B_3 \in \mathcal{B}$ such that $B_3 \subseteq (B_1 \cap B_2)$ and $x \in B_3$.

In the initially assumed scenario containing only one element in the intersection of the two elements in the basis $[x,y]$ and $[y,z]$. The only set that is possible from the intersection of these two elements from the basis is the set $\{y\}$ which has the element $y$ in it. However this set is not a part of the basis set as the basis set contains Closed Intervals $[a,b]$ where $a < b$ and this set $\{y\}$ is not such a closed interval. This scenario does not satisfy our definition of a basis for a topology. Hence the given set $\mathcal{B}$ is not a valid basis for a topology on $\mathbb{R}$.


 

### Question 2
An **open map** is a mapping between two topological spaces,
  $f : X \rightarrow Y$, such that for any open set $U \subseteq X$, the image
  $f(U)$ is an open set of $Y$.
1. Let $X = \{a, b, c\}$ and $Y = \{a, b, c\}$ both represent the three
    point set example from lecture but with potentially different topologies,
    $\mathcal{T}_X$ and $\mathcal{T}_Y$, respectively. Construct an example of
    topologies $\mathcal{T}_X$ and $\mathcal{T}_Y$ along with a mapping
    $f: X \rightarrow Y$ that is continuous but not open.
2. With the same sets $X$ and $Y$ as in part (a), construct topologies and
    a mapping ${f:X \rightarrow Y}$ that is open but not continuous.
3. Give an example of a real-valued function
    $f : \mathbb{R} \rightarrow \mathbb{R}$ that is continuous but not open.
4. Give an example of a real-valued function
    $f : \mathbb{R} \rightarrow \mathbb{R}$ that is open but not
    continuous. **Hint** You'll find such a function is quite difficult to
    come up with. You are certainly welcome to use a web search and/or AI to
    answer this question. Make sure to verify the example is correct and cite
    your sources!

### Solution
#### First Question:

Let us consider two topologies
$\mathcal{T}_x = \{\{\phi\},\{a,b,c\},\{a\}\}$ and $\mathcal{T}_y = \{\{\phi\},\{a,b,c\}\}$, These are two valid topologies on the sets $X$ and $Y$ respectively.

Choosing a mapping $f:X \rightarrow Y$ such that $f(a) = a,f(b) = b,f(c) = c$  

This mapping is continuous as both $\{\phi\}$ and $\{a,b,c\}$ have a pre-image of $\{\phi\}$ and $\{a,b,c\}$ respectively which are open sets in the topology $\mathcal{T}_X$. However, this mapping is not open as the set $\{a\}$ is an open set in the topology $\mathcal{T}_X$ but its image $f(\{a\}) = \{a\}$ is not an open set in the topology $\mathcal{T}_Y$.

#### Second Question:

Let us consider two topologies
$\mathcal{T}_x = \{\{\phi\},\{a,b,c\}\}$ and $\mathcal{T}_y = \{\{\phi\},\{a,b,c\},\{a\}\}$

Choosing a mapping $f:X \rightarrow Y$ such that $f(a) = b,f(b) = a,f(c) = c$  

This mapping is open as both $\{\phi\}$ and $\{a,b,c\}$ are open sets in the topology $\mathcal{T}_X$ and their images are $\{\phi\}$ and $\{a,b,c\}$ respectively which are open sets in the topology $\mathcal{T}_Y$. However, this mapping is not continuous as the set $\{a\}$ is an open set in the topology $\mathcal{T}_Y$ but its pre-image $f^{-1}(\{a\}) = \{a\}$ is not an open set in the topology $\mathcal{T}_X$.
