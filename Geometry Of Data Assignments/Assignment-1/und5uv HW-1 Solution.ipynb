### Question 1
Consider the set of all **closed** intervals in the real line with
  non-zero length:
  $$\mathcal{B} = \{[a, b] : a < b\}.$$
  Is this a valid basis for a topology on $\mathbb{R}$? Why or why not?



#### Solution
The given set $\mathcal{B}$ is not a valid basis for a topology on $\mathbb{R}$.

Proof : let's consider the following two intervals $[x,y]$ and $[y,z]$ where $x < y < z$ from the given Basis $\mathcal{B}$. in this scenario, the intersection of these two intervals has only one element $y$
  
From the definition of a **basis** for a topology, If $B_1, B_2 \in \mathcal{B}$ and $x \in (B_1 \cap B_2)$, then there exists a $B_3 \in \mathcal{B}$ such that $B_3 \subseteq (B_1 \cap B_2)$ and $x \in B_3$.

In the initially assumed scenario containing only one element in the intersection of the two elements in the basis $[x,y]$ and $[y,z]$. The only set that is possible from the intersection of these two elements from the basis is the set $\{y\}$ which has the element $y$ in it. However this set is not a part of the basis set as the basis set contains Closed Intervals $[a,b]$ where $a < b$ and this set $\{y\}$ is not such a closed interval. This scenario does not satisfy our definition of a basis for a topology. Hence the given set $\mathcal{B}$ is not a valid basis for a topology on $\mathbb{R}$.

### Question 3
Write a Python function to compute the stereographic projection,
  $f: S^2 \rightarrow \mathbb{R}^2$, from the north pole $(0,0,1)$ of the
  2-sphere to the $z = 0$ plane. You may want to see the Wikipedia page for the
  equation:
    https://en.wikipedia.org/wiki/Stereographic_projection

```python
# Solution
# Stereographic projection function
def stereographic_projection(point):
    x, y, z = point
    if z == 1:  
        return None # North pole is not projected since it is mapped to infinity
    X = x / (1 - z)
    Y = y / (1 - z)
    return (X, Y)
```

What is the image of a longitudinal line, i.e., a great circle passing through the north and south poles? Use your function to plot several images of logitudinal lines in the plane.

```python
# Solution
# Assisted With Code From ChatGPT and GitHub Copilot
import numpy as np
import matplotlib.pyplot as plt

# Function to generate points on a longitudinal line for a fixed phi
def longitudinal_line(phi, num_points=100):
    theta_values = np.linspace(0, np.pi, num_points)  # Vary theta from 0 to pi
    points_on_sphere = [(np.sin(theta) * np.cos(phi), 
                         np.sin(theta) * np.sin(phi), 
                         np.cos(theta)) for theta in theta_values]
    return points_on_sphere

# Plotting several longitudinal lines
phis = np.linspace(0, 2 * np.pi, 8)  # Different longitudes (8 lines)
plt.figure(figsize=(6, 6))

for phi in phis:
    points_on_line = longitudinal_line(phi)
    projected_line = [stereographic_projection(point) for point in points_on_line if stereographic_projection(point) is not None]  # Remove points at the north pole
    # Extract X and Y for plotting
    X, Y = zip(*projected_line)
    plt.plot(X, Y, label=f'phi={phi:.2f}')
    
# Plot settings
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Stereographic Projection of Longitudinal Lines (without North Pole)')
plt.legend()
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

What is the image of a latitude line, i.e., a circle formed by intersection with a horizontal plane (z = const.)? Plot several images of latitude lines in the plane.

```python
# Solution
# Assisted With Code From ChatGPT and GitHub Copilot

# Function to generate points on a latitude line for a fixed theta
def latitude_line(theta, num_points=100):
    phi_values = np.linspace(0, 2 * np.pi, num_points)  # Vary phi from 0 to 2pi
    points_on_sphere = [(np.sin(theta) * np.cos(phi), 
                         np.sin(theta) * np.sin(phi), 
                         np.cos(theta)) for phi in phi_values]
    return points_on_sphere

# Plotting several latitude lines
thetas = np.linspace(0.1, np.pi - 0.1, 6)  # Different latitudes (6 lines)
plt.figure(figsize=(6, 6))

for theta in thetas:
    points_on_line = latitude_line(theta)
    projected_line = [stereographic_projection(point) for point in points_on_line]
    
    # Filter out None values (undefined projections)
    projected_line = [point for point in projected_line if point is not None]
    
    # Extract X and Y for plotting
    X, Y = zip(*projected_line)
    plt.plot(X, Y, label=f'theta={theta:.2f}')

# Plot settings
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Stereographic Projection of Latitude Lines')
plt.legend()
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

What is the image of an arbitrary great circle (one that is not a longitudinal line or the equator)? Again, plot several in the plane.

```python
# Solution
# Assisted With Code From ChatGPT and GitHub Copilot

# Function to rotate points on the sphere by a rotation matrix
def rotate_point(point, axis, angle):
    c = np.cos(angle)
    s = np.sin(angle)
    x, y, z = point
    if axis == 'x':
        return (x, c*y - s*z, s*y + c*z)
    elif axis == 'y':
        return (c*x + s*z, y, -s*x + c*z)
    elif axis == 'z':
        return (c*x - s*y, s*x + c*y, z)

# Function to generate an arbitrary great circle on the sphere
def arbitrary_great_circle(rotation_angle, axis='x', num_points=1000):
    phi_values = np.linspace(0, 2 * np.pi, num_points)  # Vary phi from 0 to 2pi
    # Start with an equatorial great circle (theta = pi/2)
    points_on_equator = [(np.cos(phi), np.sin(phi), 0) for phi in phi_values]
    # Rotate the equator to generate an arbitrary great circle
    rotated_points = [rotate_point(point, axis, rotation_angle) for point in points_on_equator]
    return rotated_points

# Plotting several arbitrary great circles
angles = np.linspace(0, np.pi/2, 6)  # Rotate by different angles (6 great circles)
plt.figure(figsize=(6, 6))

for angle in angles[:5]:
    points_on_circle = arbitrary_great_circle(angle, axis='y')  # Rotate around the y-axis
    projected_circle = [stereographic_projection(point) for point in points_on_circle]
    
    # Filter out None values (undefined projections)
    projected_circle = [point for point in projected_circle if point is not None]
    
    # Extract X and Y for plotting
    X, Y = zip(*projected_circle)
    plt.plot(X, Y, label=f'angle={angle:.2f} rad')

# Plot settings
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Stereographic Projection of Arbitrary Great Circles')
plt.legend()
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

### Question 4

Write a Python function to sample uniformly random points on the
  $d$-sphere, $S^d$. Do this by first sampling $d+1$ i.i.d. standard Gaussian
  random variables, $N(0,1)$, to get a point
  $x = (x_1, x_2, \ldots, x_{d+1}) \in \mathbb{R}^{d+1}$. Then project this
  point to the unit sphere by $x \mapsto \frac{x}{\|x\|}$.  
  
Repeat the following experiments for $d = 2, 4, 10, 100, 1000$

```python
# Solution
# Assisted With Code From ChatGPT and GitHub Copilot

import numpy as np

def sample_points_on_sphere(d, n_samples):
    # Sample (d + 1) i.i.d. standard Gaussian random variables
    x = np.random.normal(size=(n_samples, d + 1))
    # Project onto the unit sphere
    x /= np.linalg.norm(x, axis=1, keepdims=True)
    return x
```

Sample 10,000 points on $S^d$ and compute their distance from the
    ``equator'', or great circle with $x_1 = 0$. Distance to the equator should
    be arc distance along the sphere. For each value of $d$, plot a histogram of
    these distances. What do you notice as $d$ increases? What does this say
    about uniform random points on the sphere? Would this result be the same or
    different if you used a different great circle (possibly not axis-aligned)? ****

```python
# Solution
# Assisted With Code From ChatGPT and GitHub Copilot

import matplotlib.pyplot as plt

dimensions = [2, 4, 10, 100, 1000]
n_samples = 10000

for d in dimensions:
    # Sample points on S^d
    x = sample_points_on_sphere(d, n_samples)
    # Compute distances from the equator
    theta = np.arcsin(np.abs(x[:, 0]))
    # theta is the angle between the line connecting the point and the equatorial plane
    # distance = radius * theta, where radius = 1
    # distance = theta
    # Plot histogram
    plt.figure()
    plt.hist(theta, bins=100, density=True, alpha=0.7)
    plt.title(f'Distance from Equator for d = {d}')
    plt.xlabel('Arc Distance From The Equator')
    plt.ylabel('Density')
    plt.show()
```

##### Observations:
As  $d$  increases, the histogram becomes more peaked near zero. This indicates that the distances from the random points to the equator  $x_1 = 0$  tend to decrease.  

In high-dimensional spheres, the majority of the surface area concentrates near the equator of any given axis. This phenomenon is due to the geometry of high-dimensional spaces, where the volume (or surface area) tends to concentrate in regions equidistant from the center along all dimensions.  

This result would be the same if we used a different great circle (not necessarily axis-aligned). The sphere’s symmetry ensures that any fixed equator will exhibit the same behavior. There are simply far more points near the equator than near the poles in high-dimensional spaces.

Sample 1,000 pairs of points on $S^d$ and compute the angle between
    them. Again, for each value of $d$, plot a histogram of these pairwise
    angles. What do you notice as $d$ increases and what does it say about the
    geometry of uniformly random points?

```python
n_pairs = 1000

for d in dimensions:
    # Sample pairs of points on S^d
    x = sample_points_on_sphere(d, n_pairs)
    y = sample_points_on_sphere(d, n_pairs)
    # Compute angles between pairs
    # cos(theta) = x.y / (||x|| ||y||)
    cos_theta = np.sum(x * y, axis=1)
    # theta = arccos(x.y / (||x|| ||y||))
    # We can ignore the denominator as the points are on the unit sphere and ||x|| = ||y|| = 1
    theta = np.arccos(cos_theta)
    # Plot histogram
    plt.figure()
    plt.hist(theta, bins=100, density=True, alpha=0.7)
    plt.title(f'Angle Between Pairs for d = {d}')
    plt.xlabel('Angle (radians)')
    plt.ylabel('Density')
    plt.show()
```

As  $d$  increases, the histogram becomes more peaked around  $\frac{\pi}{2}$ . This indicates that the angles between random pairs of points tend to concentrate around 90 degrees. 

in high-dimensional spaces, random vectors are almost always nearly orthogonal. This is a well-known phenomenon in high-dimensional geometry and statistics, reflecting the fact that there’s “more room” in higher dimensions, causing random vectors to spread out in such a way that they are approximately perpendicular to each other.
