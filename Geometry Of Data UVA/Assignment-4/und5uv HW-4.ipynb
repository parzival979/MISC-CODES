#### Problem 1 - Part(D) PDFs of N(0,1) and N(0,2) And Geodesic Plots


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters
mu1, sigma1 = 0, 1
mu2, sigma2 = 0, 2

# Plot range
x = np.linspace(-10, 10, 1000)

# PDFs
pdf1 = norm.pdf(x, mu1, sigma1)
pdf2 = norm.pdf(x, mu2, sigma2)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(x, pdf1, label='N(0, 1)')
plt.plot(x, pdf2, label='N(0, 2)')
plt.title('PDFs of N(0,1) and N(0,2)')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Parameters for the geodesic path
a = 0  # Mu is constant
b = 1  # Starting sigma
c = np.log(2)  # Natural log of 2
t = np.linspace(0, 1, 100)  # Parameter for the geodesic curve

# Geodesic path
mu_t = np.full_like(t, a)  # Mu is constant
sigma_t = b * np.exp(c * t)  # Sigma varies exponentially

# Plot in the Poincaré half-plane
plt.figure(figsize=(8, 6))
plt.plot(mu_t, sigma_t, label='Geodesic Path', color='blue')
plt.scatter([a, a], [b, b * np.exp(c)], color='red', zorder=5, label='Endpoints')
plt.title('Geodesic in the Poincaré Half-Plane from (0,1) to (0,2)')
plt.xlabel('$\mu$')
plt.ylabel('$\sigma$')
plt.legend()
plt.grid(True)
plt.show()
```

## Problem 2
Write two functions to compute the maximum-likelihood estimate (MLE) of a logistic regression from training data. In the first function, use standard gradient descent with a fixed
step size. In the second function, use the natural gradient with an empirical estimate for the
Fisher information matrix.

You should implement the gradient descent of logistic regression yourself, rather than just
use a library, to make sure that you know what you are doing. However, feel free to look up
the equations for the log-likelihood (cross-entropy) gradient on the internet.

(a) Pick a data set (suggestions: MNIST, Fashion MNIST, CIFAR-10), and pick two classes
from that data set to test your binary classifier. You may use a suitable subsample of
the training set to save time.

(b) Run your fixed-step gradient descent on the training data. Use a random initialization
for the weight vector, and experiment to find a step size that works reasonably well. Run
for a fixed number of iterations (at least 10). Save the training loss for each iteration.

(c) Using the same random initialization for the weight vector that you used above, run
your natural gradient descent method for the same number of iterations. Again, save
the training loss at each iteration.

(d) Repeat the previous two steps several times (at least 5 times). Each repeat should
use a new random weight initialization, but always equal for the two algorithms. Plot
the average loss as a function of the iteration for both algorithms. Which algorithm
converges faster?

```python
# Reference: https://courses.cs.washington.edu/courses/cse599i/18wi/resources/lecture8/lecture8.pdf Page 5

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import StandardScaler

# Load and preprocess the MNIST dataset
def load_mnist_digits(digit1=0, digit2=1, sample_size=2000):
    # Fetch MNIST dataset
    mnist = fetch_openml('mnist_784', version=1, as_frame=False)
    X = mnist['data']
    y = mnist['target'].astype(np.int8)

    # Filter for the two digits
    filter_digits = (y == digit1) | (y == digit2)
    X, y = X[filter_digits], y[filter_digits]
    y = (y == digit2).astype(np.int8)  # Convert to binary labels (0 and 1)

    # Subsample the data by selecting a random subset
    np.random.seed(0)
    idx = np.random.choice(len(y), sample_size, replace=False)
    X, y = X[idx], y[idx]

    # Standardize features for better learning
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Add bias term to the features to account for the intercept term
    X_scaled = np.hstack([np.ones((X_scaled.shape[0], 1)), X_scaled])

    return X_scaled, y

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Compute binary cross-entropy loss
def compute_loss(y, y_pred):
    epsilon = 1e-7  # To prevent log(0)
    loss = -np.mean(y * np.log(y_pred + epsilon) + (1 - y) * np.log(1 - y_pred + epsilon))
    return loss

# Standard Gradient Descent
def standard_gradient_descent(X, y, w_init, learning_rate=0.01, iterations=10):
    w = w_init.copy()
    losses = []
    for i in range(iterations):
        # Forward pass, We multiply X with w to get the linear combination of features and weights, Then we apply the sigmoid function to get the probability of the data point belonging to class
        z = X @ w
        y_pred = sigmoid(z)

        # Compute loss
        loss = compute_loss(y, y_pred)
        losses.append(loss)

        # Gradient computation with Derivative of the loss function with respect to the weights, and we divide by the number of samples because of the mean in the loss function
        gradient = X.T @ (y_pred - y) / len(y)

        # Parameter update
        w -= learning_rate * gradient

    return w, losses

# Natural Gradient Descent
def natural_gradient_descent(X, y, w_init, learning_rate=0.1, iterations=10):
    w = w_init.copy()
    losses = []
    for i in range(iterations):
        # Forward pass
        z = X @ w
        y_pred = sigmoid(z)

        # Compute loss
        loss = compute_loss(y, y_pred)
        losses.append(loss)

        # Gradient computation
        gradient = X.T @ (y_pred - y) / len(y)

        # Compute empirical Fisher Information Matrix
        D = np.diag(y_pred * (1 - y_pred))
        Fisher_matrix = (X.T @ D @ X) / len(y)

        # Add a small value to diagonal for numerical stability
        epsilon = 1e-7
        Fisher_matrix += epsilon * np.eye(Fisher_matrix.shape[0])

        # Natural gradient update
        # Here, we solve a linear system to compute the natural gradient (Fisher_matrix @ nat_gradient = gradient) instead of directly computing the inverse, This gives nat_gradient = inv(Fisher_matrix) @ gradient
        nat_gradient = np.linalg.solve(Fisher_matrix, gradient)
        w -= learning_rate * nat_gradient

    return w, losses

# Run experiments
def run_experiments(X, y, learning_rate=0.1, iterations=10, repeats=5):
    losses_sgd_all = []
    losses_ngd_all = []

    for i in range(repeats):
        # Initialize weights randomly but the same for both algorithms
        np.random.seed(i)
        w_init = np.random.randn(X.shape[1])

        # Standard Gradient Descent
        _, losses_sgd = standard_gradient_descent(X, y, w_init, learning_rate, iterations)
        losses_sgd_all.append(losses_sgd)

        # Natural Gradient Descent
        _, losses_ngd = natural_gradient_descent(X, y, w_init, learning_rate, iterations)
        losses_ngd_all.append(losses_ngd)

    # Compute average losses
    avg_losses_sgd = np.mean(losses_sgd_all, axis=0)
    avg_losses_ngd = np.mean(losses_ngd_all, axis=0)

    return avg_losses_sgd, avg_losses_ngd

# Main execution
if __name__ == "__main__":
    # Load data for digits 0 and 1
    X, y = load_mnist_digits(digit1=0, digit2=1, sample_size=2000)

    # Set parameters
    learning_rate = 0.1
    iterations = 25
    repeats = 10

    # Run experiments
    avg_losses_sgd, avg_losses_ngd = run_experiments(X, y, learning_rate, iterations, repeats)

    # Plot average losses
    plt.figure(figsize=(10, 6))
    plt.plot(range(1, iterations + 1), avg_losses_sgd, marker='o', label='Standard Gradient Descent')
    plt.plot(range(1, iterations + 1), avg_losses_ngd, marker='s', label='Natural Gradient Descent')
    plt.xlabel('Iteration')
    plt.ylabel('Average Training Loss')
    plt.title('Average Training Loss vs Iterations')
    plt.legend()
    plt.grid(True)
    plt.show()
```

#### From the plot, we can see that the Natural Gradient Descent converges faster than the Standard Gradient Descent. The Natural Gradient Descent uses the Fisher Information Matrix to compute the natural gradient, which is more efficient than the Standard Gradient Descent, since it takes into account the nature of the loss.
